```python
def extract_episode_count(text):
    #send all titles to this function
    check=False
    data=""
    for i in text:
        if i==')':
            break
        if i=='(':
            check=True
            continue
        if check==True:
            data+=i
    return data
```

```python
def extract_timestamp(text):
    check=False
    data=""
    for i in range(len(text)):
        if text[i]==')':
            for j in range(i+1,i+20):
                data+=text[j]
            break
    return data


```

```python
df['timestamp']=df['Title'].apply(extract_timestamp)
df


#months extract to do
#python less fucntions
#total time se months extract


```

```python
from datetime import datetime

def extract_total_months(date_range):
    # Example: "Apr 2009 - Jul 2010"
    if ' - ' not in date_range:
        return 0
    
    try:
        # Split by ' - '
        start_str, end_str = date_range.split(' - ')
        
        # Parse dates
        start_date = datetime.strptime(start_str.strip(), '%b %Y')
        end_date = datetime.strptime(end_str.strip(), '%b %Y')
        
        # Calculate months
        months = (end_date.year - start_date.year) * 12 + \
                 (end_date.month - start_date.month)
        
        return months
    except:
        return 0

# Test
print(extract_total_months("Apr 2009 - Jul 2010"))  # 15

# Use with Pandas
df['Total_Months'] = df['timestamp'].apply(extract_total_months)
```
