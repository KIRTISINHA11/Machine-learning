```python
words=["eating","eaten","eats", "writting", "written", "writes"]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word, " : ", stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

```python
stemming.stem("sitting")
```

#snowballstemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstememr=SnowballStemmer("english")
```

```python
for word in words:
    print(word, " : ", snowballstememr.stem(word))
```
